**Грубость неустойчивой точки равновесия маятника** $$\begin{cases} \dot{x}_2=-\sin x_1, \\ \dot{x}_1=x_2 \end{cases}$$

```python
import matplotlib.pyplot as plt
from numpy import sin,pi,linspace

def plotvec(ax, x1, x2, d=0.1, perturbate=False):
    dx1 = x2 * d
    dx2 = -sin(x1) * d
    if perturbate:
        dx1 += x2*0.07
        dx2 += -(x1+pi)*0.07
    ax.plot([x1,x1+dx1], [x2,x2+dx2], 'blue')
    ax.scatter(x1, x2, 5, 'g')

N = 20
e = 0.1
fig, axs = plt.subplots(1,2, figsize=(13,6))
for x1 in linspace(-pi-e,-pi+e,N):
    for x2 in linspace(-e,e,N):
        plotvec(axs[0], x1, x2)
        plotvec(axs[1], x1, x2, perturbate=True)
axs[0].set_title("Невозмущённое седло")
axs[1].set_title("Возмущённое седло")
plt.show()
```

**Бифуркация системы** $$\begin{cases} \dot x=\mu - (x+1)(x-1) \\ \dot y=-y \end{cases}$$

```python
import matplotlib.pyplot as plt
from numpy import sin,pi,linspace

def plotvec(ax, x, y, mu, d=0.01):
    x_ = [x]
    y_ = [y]
    for i in range(10):
        x_.append(x_[-1] + ((mu - (x_[-1]+1)*(x_[-1]-1))) * d)
        y_.append(y_[-1] + (-y_[-1]) * d)
    ax.plot(x_, y_, 'blue')
    ax.scatter(x, y, 5, 'g')

N = 15
e = 10
fig, axs = plt.subplots(1,3, figsize=(13,4))
for x in linspace(-e,+e,N):
    for y in linspace(-e,e,N):
        plotvec(axs[0], x, y, -5)
        plotvec(axs[1], x, y, -1)
        plotvec(axs[2], x, y,  3)
axs[0].set_title("μ=-5")
axs[1].set_title("μ=-1")
axs[2].set_title("μ=3")
for i in range(3):
    axs[i].set_xlim([-e, e])
plt.show()
```

**Бифуркация системы** $$\dot x=\mu_1+\mu_2 x^2 - x^4$$

```python
from sympy import *
import matplotlib.pyplot as plt
from numpy import sin,pi,linspace
x = var('x', real=True)

N = 35
e = 10
plt.figure(figsize=(10,5))
for mu1 in linspace(-e,e,N):
    for mu2 in linspace(-e,e,N):
        n = len(solve(Eq(0, mu1 + mu2*x**2 - x**4), x))
        plt.scatter(mu1, mu2, 20, 'kcmgyr'[n])

for i in range(5):
    plt.scatter(15, 10-i/4, 20, 'kcmgy'[i], label = ['0 равновесий','1 равновесие','2 равновесия','3 равновесия','4 равновесия'][i])
plt.axis('equal')
plt.legend(loc='upper left')
plt.xlabel('μ1')
plt.ylabel('μ2')
plt.show()
```
