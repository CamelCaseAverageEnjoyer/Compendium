```python
from sympy import *

x, y, r, v, t, m = var('x y r v t mu')
r = Function('r')(t)
v = Function('v')(t)
x = r + v
y = r - v
eqs = [Eq(x.diff(t),m*x-x**2), Eq(y.diff(t),-y)]
anw = solve(eqs, [r.diff(t), v.diff(t)])
display(Eq(r.diff(t),anw[r.diff(t)].simplify()))
display(Eq(v.diff(t),anw[v.diff(t)]))
```
